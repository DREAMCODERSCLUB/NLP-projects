```python
import re
import nltk
from nltk.corpus import stopwords
```

```python
# Download stopwords if not already downloaded
nltk.download('stopwords')
```

```python
# Load sample text (can be replaced with file read)
sample_text = """
Natural Language Processing (NLP) is a sub-field of Artificial Intelligence.
It deals with the interaction between computers and humans using natural language.
"""
```

```python
# Step 1: Lowercase
text = sample_text.lower()
```

```python
print(text)
```

```python
# Step 2: Remove punctuation and special characters
text = re.sub(r'[^a-z\s]', '', text)
print(text)
```

```python
# Step 3: Tokenize (split by spaces)
tokens = text.split()
```

```python
# Step 4: Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in stop_words]
```

```python
# Output
print("Original Tokens:", tokens)
print("Filtered Tokens:", filtered_tokens)
```
